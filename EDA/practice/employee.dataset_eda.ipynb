### EDA - Employee Dataset


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline 
```

```python
df = pd.read_csv('Cleaned_Employee_Dataset.csv')
df.head()
```

#### Histogram : To see how neumaric value is distributed

```python
df.hist(bins=10,figsize=(20,15))
plt.show()
```

#### Box Plot : To identify outliers and show the five-number summary

```python
df.plot(kind='box',subplots=True,layout=(4,4),figsize=(20,15))
plt.show()
```

#### Bar charts : To see category frequencies

```python
department_counts = df['Department'].value_counts()
plt.bar(department_counts.index, department_counts.values, color='skyblue')
plt.xlabel('Department')
plt.ylabel('Count')
plt.title('Count of Employees by Department')
plt.show()
```

```python
sns.countplot(x='Remote_Work', data=df)
plt.show()
```

```python
sns.countplot(x='Performance_Score', data=df)
plt.show()
```

```python
# use countplots to see category frequencies.
sns.countplot(x='Status', data=df)
plt.show()
```

#### Correlation : To identify relationships between variables

```python
plt.figure(figsize=(12,6))
sns.heatmap(df.corr(numeric_only=True))
plt.show()
```

#### Scatter/pair plot : To identify linear or non-linear relationships.

```python
# Use a scatter plot to identify linear or non-linear relationships
sns.pairplot(
    df[["Age", "Salary", "Performance_Score"]],
    kind="scatter",
    diag_kind="kde",
)
plt.show()
```

#### Pie Charts : To showing the percentage distribution of categories

```python
department_counts = df['Department'].value_counts()
plt.pie(department_counts, labels= department_counts.index, autopct = "%1.1f%%",startangle=90)
plt.title('Employee Distribution by Department')
plt.show()
```

```python
df['Age'].value_counts().plot(kind='pie', color='lightgreen',autopct = "%1.1f%%",startangle=90)
plt.show()
```

```python
df['Region'].value_counts().plot(kind='pie', autopct='%1.1f%%', startangle=90)
plt.show()
```

#### Bar plot : Use to see one feature with another

```python
grouped = df.groupby('Department')['Salary'].mean()
sns.barplot(x=grouped.index, y=grouped.values)
plt.show()
```

```python
grouped = df.groupby('Department')['Performance_Score'].value_counts().unstack()
grouped.plot(kind="bar", stacked=True)
plt.show()
```

```python
grouped = df.groupby('Region')['Salary'].mean()
sns.barplot(x=grouped.index, y=grouped.values)
plt.show()
```
