14.Create a new column called quality_label:
- If quality ≥ 7 → "Good"
- If quality = 5 or 6 → "Average"
- If quality ≤ 4 → "Poor"

```python
def label_quality(q):
    if q >=7:
        return 'Good'
    elif q == 5 or q == 6:
        return 'Average'
    else:
        return 'Poor'

df['quality_label'] = df['quality'].apply(label_quality)
```

15. Count how many wines fall into each quality label.

16. Find the average alcohol content for each quality category.
