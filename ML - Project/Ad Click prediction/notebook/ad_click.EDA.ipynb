### Ad Click prediction - EDA

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
df = pd.read_csv(r'C:\Upskill\Data Science\ML - Project\Ad Click prediction\data\cleaned_ad_10000records.csv')
df.head()
```

```python
df.hist(bins=10,figsize=(20,15))
plt.show()
```

```python
df.info()
```

```python
TOS_count = df['Time on Site'].value_counts()
plt.bar(TOS_count.index, TOS_count.values, color='skyblue')
plt.show()
```

```python
Age_count = df['Age'].value_counts()
plt.bar(Age_count.index, Age_count.values, color='blue')
plt.show()
```

```python
gender_counts = df['Gender'].value_counts()
plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%', startangle=90)
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.title('Gender Distribution')
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True),annot=True, cmap='coolwarm')
plt.show()
```

```python
df.info()
```

```python
grouped = df.groupby('Clicked on Ad')['Age'].mean()
plt.bar(grouped.index, grouped.values, color='green')
plt.xlabel('Clicked on Ad')
plt.ylabel('Average Age')
plt.title('Average Age by Ad Click Status')
plt.show()
```

```python
df['Gender'] = df['Gender'].map({'Male': 0, 'Female': 1})
```
