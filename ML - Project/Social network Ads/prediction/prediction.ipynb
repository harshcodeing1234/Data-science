```python
import pickle
import numpy as np
```

```python
with open(r"C:\Upskill\Data Science\ML - Project\Social network Ads\model.pkl", "rb") as f:
    bundle = pickle.load(f)

model = bundle['model']
scaler = bundle['scaler']

gender = int(input("Enter Gender (Male=1, Female=0): "))
age = int(input("Enter Age: "))
salary = int(input("Enter Salary: "))
new_data = [[gender, age, salary]]

new_data_scaled = scaler.transform(new_data)

prediction = model.predict(new_data_scaled)

if prediction[0] == 1:
    print("Customer will PURCHASE") 
else:
    print("Customer will NOT purchase")
```
