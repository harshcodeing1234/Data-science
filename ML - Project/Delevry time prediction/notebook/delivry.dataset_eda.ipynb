### Delivery Dataset EDA 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
df = pd.read_csv(r"C:\Upskill\Data Science\ML - Project\Delevry time prediction\data\Food_Delivery_Times_cleaned.csv")
df.head()
```

```python
df.hist(bins=10,figsize=(20,15))
plt.show()
```

```python
df.plot(kind='box',subplots=True,layout=(3,5),figsize=(20,15))
plt.show()
```

```python
courier_exprience_counts = df["Courier_Experience_yrs"].value_counts()
plt.bar(courier_exprience_counts.index,courier_exprience_counts.values,color="orange")
plt.title("Courier Experience frequency")
plt.xlabel("Courier Experience")
plt.ylabel("Count")
plt.show()
```

```python
traffic_counts = df["Traffic_Level"].value_counts()
plt.bar(traffic_counts.index,traffic_counts.values,color="green")
plt.title("Traffic Level frequency")
plt.xlabel("Traffic Level")
plt.ylabel("Count")
plt.show()
```

```python
vehical_counts = df["Vehicle_Type"].value_counts()
plt.bar(vehical_counts.index,vehical_counts.values,color="red")
plt.title("Vehicle Type frequency")
plt.xlabel("Vehicle Type") 
plt.ylabel("Count")
plt.show()
```

```python
weather_counts = df["Weather"].value_counts()
plt.bar(weather_counts.index,weather_counts.values,color="skyblue")
plt.title("Weather frequency")
plt.xlabel("Weather")
plt.ylabel("Count")
plt.show()
```

```python
# correlation matrix
plt.figure(figsize=(10,8))
plt.title("Correlation Matrix")
sns.heatmap(df.corr(numeric_only=True),annot=True,cmap="coolwarm",linewidths=0.5)
plt.show()
```

```python
grouped_data = df.groupby('Weather')['Delivery_Time_min'].mean().reset_index()
plt.figure(figsize=(10,6))
sns.barplot(x='Weather', y='Delivery_Time_min', data=grouped_data, palette='viridis')
plt.title('Average Delivery Time by Weather Condition')
plt.show()
```
