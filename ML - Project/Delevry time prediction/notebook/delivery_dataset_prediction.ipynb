### Delivery time prediction 


```python
import pandas as pd
```

```python
# load model 
import joblib

model = joblib.load("C:\\Upskill\\Data Science\\ML - Project\\Delevry time prediction\\models\\linear_regression_model.pkl")
scaler = joblib.load("C:\\Upskill\\Data Science\\ML - Project\\Delevry time prediction\\models\\scaler.pkl")
feature_columns = joblib.load("C:\\Upskill\\Data Science\\ML - Project\\Delevry time prediction\\models\\feature_columns.pkl")
```

```python
# give input data for prediction
new_data = pd.DataFrame({
    "Distance_km": [10],
    "Weather": ["Rainy"],
    "Traffic_Level": ["High"],
    "Time_of_Day": ["Evening"],
    "Vehicle_Type": ["Bike"],
    "Preparation_Time_min": [15],
    "Courier_Experience_yrs": [2]
})
```

```python
# encode input data using one-hot encoding
new_data = pd.get_dummies(
    new_data,
    columns=["Weather", "Traffic_Level", "Time_of_Day", "Vehicle_Type"],
    dtype=int
)
```

```python
# arrange column according training columns 
new_data = new_data.reindex(
    columns=feature_columns,
    fill_value=0
)
```

```python
# scale the input data using the same scaler used during training
new_data_scaled = scaler.transform(new_data)
```

```python
# predict delivery time using the trained model
prediction = model.predict(new_data_scaled)

print("Predicted Delivery Time:", prediction[0], "minutes")
```
