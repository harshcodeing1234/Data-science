#### Data cleaning : Employee Dataset

```python
# import libraries 
import numpy as np 
import pandas as pd
```

```python
# load dataset
df = pd.read_csv('Messy_Employee_dataset.csv')
df.head()
```

##### Data cleaning

```python
# check missing values

df.isnull().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean(),inplace = True) 

# check age dtype to int 
df['Age'] = df['Age'].astype(int)

df['Age'].dtype
```

```python
df['Salary'] = df['Salary'].fillna(df['Salary'].mean(),inplace=True)
```

```python
df.head()
```

```python
df['Salary'] = df['Salary'].astype(int)
df['Salary'].dtype
```

```python
# Remove duplicates
df.duplicated().sum() # no duplicates found
# df.drop_duplicates(inplace=True) if duplicate use this
```

```python
# standrized text
df['Employee_ID'] = df['Employee_ID'].str.replace("EMP","",regex=False)
df['Employee_ID'] = df['Employee_ID'].astype('int')
```

```python
# Combine first_name and last_name in name
df['Name'] = df['First_Name'] + " " + df['Last_Name']
```

```python
# move name column to first position
df.insert(1,"Name", df.pop("Name"))
```

```python
# drop first name and last name columns
df.drop(columns=['First_Name','Last_Name'],inplace=True)
```

```python
# split Department_Region to Depart and Region 
df[["Department", "Region"]] = df["Department_Region"].str.split("-", expand=True)
```

```python
df.drop(columns=['Department_Region'],inplace=True)
```

```python
# Fix joining date 
df['Join_Date'] = pd.to_datetime(df['Join_Date'], format='mixed')
```

```python
df['Phone'] = df['Phone'].astype('str')
df['Phone'] = df['Phone'].str.replace("-"," ",regex=False)
df['Phone'] = df['Phone'].astype('int')
```

```python
# Save this cleaned data for EDA
df.to_csv('Cleaned_Employee_dataset.csv',index=False)
```
