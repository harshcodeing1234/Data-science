Create a program that takes an electricity consumption unit and calculates the bill using different slab rates. Also add a fixed ₹100 charge if the final bill exceeds ₹2,000.

````python
def calculate_electric_bill(units):
    bill_amount = 0.00
    if units <=100:
        bill_amount = (units * 4)
    elif units <=300:
        bill_amount = (100*4) + ((units - 100) * 6)
    else:
        bill_amount = (100*4) + (200*6) + ((units - 300) * 8)

    print(f"Base bill amount for {units} units is: {bill_amount}")

    if bill_amount > 2000:
        bill_amount =  bill_amount + 100
        print(f"Surcharge applied. Total bill amount is: {bill_amount}")
    else:
        print(f"Surcharge not applied. Total bill amount is: {bill_amount}")

    return bill_amount

try:
    user_units = float(input("Enter the number of units consumed: "))
    if user_units < 0:
        print("Units consumed cannot be negative.")
    else:
        calculate_electric_bill(user_units)
except ValueError:
    print("Invalid input. Please enter a numeric value for units consumed.")
````

````python
numbers = [12, 5, 8, 20, 3, 15, 8, 20, 1]

maximum = numbers[0]
minimum = numbers[0]

total = 0
count = 0


# find maximum , minimum and total

for num in numbers:
    if num > maximum:
        maximum = num

    if num < minimum:
        minimum = num
    total += num
    count +=1

# find average
average = total/count

# find second largest unique number

second_largest_unique = None
for num in numbers:
    if num == maximum:
        continue
    if second_largest_unique is None or num > second_largest_unique:
        second_largest_unique = num

print(f"Maximum: {maximum}")
print(f"Minimum: {minimum}")
print(f"Total: {total}")
print(f"Average: {average:.2f}")
print(f"Second largest unique: {second_largest_unique}")

````

````python
# check for prime 

def is_prime(n):
    if n <=1:
        return False
    if n <=3:
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False

    # Check factors up to the square root of n
    i = 5
    while i * i <= n:
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6
    return True

print(is_prime(11))  
````

````python
# check factorial 
def factorial(n):
    if n < 0:
        return "Factorial is not defined for negative numbers."

    fact = 1
    for i in range (1, n+1):
        fact = fact * i
    return fact
print(factorial(5))
````

Create a calculator that accepts two numbers and an operator:

```python
+
-
*
/
%
```

Handle:

- Invalid input
- Division by zero
- Invalid operator

Use try, except, else, and finally appropriately.

````python
def calculator():
    try:
        num1 = float(input('Enter first number:'))
        operator = input('Enter operator (+, -, *, /): ').strip()
        num2 = float(input('ENter second number:'))

        result = None

        if operator == '+':
            result = num1 + num2
        elif operator == '-':
            result = num1 - num2
        elif operator == '*':
            result = num1 * num2
        elif operator == '/':
            if num2 == 0:
                print("Error: Division by zero is not allowed.")
                return
            result = num1 / num2
        else:
            print("Error: Invalid operator. Please use +, -, *, or /.")
            return
    except ValueError as e:
        print(e)
    except ZeroDivisionError as e:
        print(e)
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
    else:
        print(f"The result of {num1} {operator} {num2} is: {result}")
    finally:
        print("Thank you for using the calculator.")

calculator()
````

Create a BankAccount class with:

```python
account_holder
balance
```

Methods:
```python
deposit()
withdraw()
check_balance()
```

Prevent withdrawal if the user doesn't have sufficient balance.

````python
class BankAccount:
    def __init__(self, account_holder, balance=0.0):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited {amount}. New balance is {self.balance}.")
        else:
            print("Invalid deposit amount.")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds.")
        elif amount <= 0:
            print("Invalid withdrawal amount.")
        else:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance is {self.balance}.")

    def check_balance(self):
        print(f"Account Holder: {self.account_holder} | Current Balance: Rs.{self.balance:.2f}")
        return self.balance

account = BankAccount("Rishu kumar", 1000.0)
account.deposit(2000)
account.check_balance()
account.withdraw(500)
account.check_balance()
````

Create a small Student Management System.

Features:

1. Add student
2. View students
3. Search student
4. Update marks
5. Delete student
6. Exit

Store the data in a JSON file so that data remains available after restarting the program.

````python
import os
import json

Data_file = 'data.json'

def load_data():
    if not os.path.exists(Data_file):
        return {}
    try:
        with open(Data_file, 'r') as f:
            return json.load(f)
    except json.JSONDecodeError:
        print("Error: The data file is corrupted. Starting with an empty dataset.")
        return {}

def save_data(data):
    with open(Data_file, 'w') as f:
        json.dump(data, f, indent = 4)

def add_student(students):
    student_id = len(students) + 1
    name = input("Enter student name: ")
    age = int(input("Enter student age: "))
    course = input("Enter student course: ")
    Enrollment_no = student_id
    marks = float(input("Enter student marks: "))

    students[Enrollment_no] = {
        "name": name,
        "age": age,
        "course": course,
        "marks": marks
    }

    save_data(students)
    print(f"Student {name} added successfully with Enrollment No: {Enrollment_no}")

def view_students(students):
    if not students:
        print("No students found.")
        return

    for Enrollment_no, info in students.items():
        print(f"Enrollment No: {Enrollment_no}")
        print(f"Name: {info['name']}")
        print(f"Age: {info['age']}")
        print(f"Course: {info['course']}")
        print(f"Marks: {info['marks']}")
        print("-------------------------")

def search_student(students):
    Enrollment_no = int(input("Enter Enrollment No to search: "))
    student = students.get(Enrollment_no)

    if student:
        print(f"Enrollment No: {Enrollment_no}")
        print(f"Name: {student['name']}")
        print(f"Age: {student['age']}")
        print(f"Course: {student['course']}")
        print(f"Marks: {student['marks']}")
    else:
        print("Student not found.")

def update_student(students):
    Enrollment_no = int(input("Enter Enrollment No to update: "))
    student = students.get(Enrollment_no)

    if student:
        print("Leave field empty to keep current value.")
        name = input(f"Enter new name (current: {student['name']}): ") or student['name']
        age_input = input(f"Enter new age (current: {student['age']}): ")
        age = int(age_input) if age_input else student['age']
        course = input(f"Enter new course (current: {student['course']}): ") or student['course']
        marks_input = input(f"Enter new marks (current: {student['marks']}): ")
        marks = float(marks_input) if marks_input else student['marks']

        students[Enrollment_no] = {
            "name": name,
            "age": age,
            "course": course,
            "marks": marks
        }

        save_data(students)
        print(f"Student {Enrollment_no} updated successfully.")
    else:
        print("Student not found.")

def delete_student(students):
    Enrollment_no = int(input("Enter Enrollment No to delete: "))
    if Enrollment_no in students:
        del students[Enrollment_no]
        save_data(students)
        print(f"Student {Enrollment_no} deleted successfully.")
    else:
        print("Student not found.")

def main():
    students = load_data()

    while True:
        print("\nStudent Management System")
        print("1. Add Student")
        print("2. View Students")
        print("3. Search Student")
        print("4. Update Student")
        print("5. Delete Student")
        print("6. Exit")

        choice = input("Enter your choice: ")

        if choice == '1':
            add_student(students)
        elif choice == '2':
            view_students(students)
        elif choice == '3':
            search_student(students)
        elif choice == '4':
            update_student(students)
        elif choice == '5':
            delete_student(students)
        elif choice == '6':
            print("Exiting the program.")
            break
        else:
            print("Invalid choice. Please try again.")

if __name__ == "__main__":
    main()
````
