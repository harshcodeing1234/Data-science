```python
# import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load The data
df = pd.read_csv('polynomial.csv')
df.head()
```

```python
# Visualize the data
plt.scatter(df['Temperature'],df['Pressure'])
plt.show()
```

```python
# Define feature X and Target Y
x = df[['Temperature']]
y = df['Pressure']
```

```python
# Split data into Test and Train
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
# Implement Linear Regression
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
# Predict train data
predict = model.predict(x_train)
predict
```

```python
# predict test data
test_predict = model.predict(x_test)
test_predict
```

```python
plt.scatter(x_train,y_train,marker='*',label='Linear Regression')
plt.plot(x_train,predict)
plt.legend()
plt.show()
```

```python
# Implement Polynomial Regression
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=4)
x_poly = poly.fit_transform(x_train)
x_poly
```

```python
poly.fit(x_poly,y_train)
lin2 = LinearRegression()
lin2.fit(x_poly,y_train)
```

```python
# Visualize Polynomial regression
plt.scatter(x,y,marker='*',label='Polynomial Regression',color='purple')
plt.plot(x,lin2.predict(poly.fit_transform(x)))
plt.legend()
plt.show() 
```

```python
# Model Coefficient and Intercept
print(f'Regression coeff',model.coef_)
print(f'Regression intercept: "{model.intercept_}"')
```

```python
# Predict new Value
lin2.predict(poly.fit_transform([[12]]))
```
