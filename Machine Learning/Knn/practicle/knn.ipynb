#### Knn - K- nearest neighbours

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Data created
data = {
    'x1' : [3,5,2,8,4,7,5,9,7,6,1],
    'x2' : [2,4,2,3,6,6,3,4,3,5,9],
    'y'  : [1,0,0,0,1,0,1,1,0,1,1]
}

df = pd.DataFrame(data)
df.head()
```

```python
plt.scatter(df.x1,df.x2,c=df.y)
plt.xlabel('X1')
plt.ylabel('X2')
plt.show()
```

```python
x = df[['x1','x2']]
y = df['y']
```

```python
# model training..
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier(n_neighbors=1)
model.fit(x,y)
```

```python
# prediction 
prediction = model.predict([[7,5]])
print(prediction)
```

```python
prediction = model.predict([[10,3]])
print(prediction)
```

```python
prediction = model.predict([[4,7]])
print(prediction)
```

```python
prediction = model.predict([[2.5,2.5]])
print(prediction)
```

```python
prediction = model.predict([[-5,-3]])
print(prediction)
```

```python
prediction = model.predict([[100,200]])
print(prediction)
```

```python
prediction = model.predict([[1000,5]])
print(prediction)
```
