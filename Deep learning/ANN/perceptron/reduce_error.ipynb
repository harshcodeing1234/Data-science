```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('placement.csv')
df.head()
```

```python
x = df[['cgpa', 'resume_score']]
y = df['placed']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score, zero_one_loss
```

```python
model = Perceptron(eta0=0.1, random_state=42, warm_start=True)
classes = np.unique(y_train)
```

```python
for epoch in range(1,1000):
    model.partial_fit(x_train, y_train, classes=classes)

    train_pred = model.predict(x_train)
    test_pred = model.predict(x_test)

    train_loss = zero_one_loss(y_train, train_pred)
    train_error = np.sum(train_pred != y_train)
    test_accuracy = accuracy_score(y_test, test_pred)

    print(
        f"{epoch:<8}"
        f"{train_loss:<15.4f}"
        f"{train_error:<10}"
        f"{test_accuracy:<15.4f}"
    )

    if train_error == 0:
        print(f"Stopping training at epoch {epoch} as training error is zero.")
        break    
```

```python
weight = model.coef_[0]
bias = model.intercept_[0]

print(f"Weight: {weight}, Bias: {bias}")
```

```python
new_data = np.array([
    [2.0, 1.5],
    [-1.0, -2.0],
    [0.5, 1.0]
])

prediction = model.predict(new_data)
print(f"Predictions for new data: {prediction}")
```
